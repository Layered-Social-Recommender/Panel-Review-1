```python
#create edges.csv
edges = trust.copy()
edges = edges.rename(columns={"source":"Source","target":"Target"})
```

```python
edges
```

```python
edges["Relationship"] = edges["value"].map({1:"Trust",-1:"Distrust"})
```

```python
edges["Type"] = "Directed"
```

```python
edges["Weight"] = 1
```

```python
edges
```

```python
edges = edges[["Source","Target","Type","Weight","Relationship"]]
```

```python
edges
```

```python
edges.to_csv("edges.csv",index=False)
```

```python
#most active users

degree = pd.concat([edges["Source"],edges["Target"]]).value_counts()
degree
```

```python
top_users = degree.head(5000).index
```

```python
sample_edges = edges[
    edges["Source"].isin(top_users) &
    edges["Target"].isin(top_users)
]
```

```python
sample_edges = sample_edges.sample(
    n=100000,
    random_state=42
)
```

```python
sample_nodes = pd.DataFrame({
    "Id": pd.unique(
        pd.concat([
            sample_edges["Source"],
            sample_edges["Target"]
        ])
    )
})
```

```python
sample_nodes["Label"] = sample_nodes["Id"]

sample_nodes.to_csv("sample_nodes.csv", index=False)
sample_edges.to_csv("sample_edges.csv", index=False)
```
